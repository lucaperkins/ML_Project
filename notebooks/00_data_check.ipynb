```python
from datasets import load_dataset
import pandas as pd
from sklearn.preprocessing import MultiLabelBinarizer

ds_simplified = load_dataset("google-research-datasets/go_emotions", "simplified")
label_names = ds_simplified["train"].features["labels"].feature.names

print("Number of labels:", len(label_names))
print("Sizes:", len(ds_simplified["train"]), len(ds_simplified["validation"]), len(ds_simplified["test"]))

df_train = ds_simplified["train"].to_pandas()
df_val = ds_simplified["validation"].to_pandas()
df_test = ds_simplified["test"].to_pandas()

# multi-hot encoding: one 0/1 column per label
mlb = MultiLabelBinarizer(classes = list(range(len(label_names))))
y_train = mlb.fit_transform(df_train["labels"])
y_val = mlb.transform(df_val["labels"])
y_test = mlb.transform(df_test["labels"])

# class imbalance
freq_train = pd.Series(y_train.sum(axis = 0), index = label_names).sort_values(ascending = False)
print(freq_train)

# how many labels per comment
nb_labels_train = y_train.sum(axis = 1)
print("Share of comments with one label:", (nb_labels_train == 1).mean())

# leakage check: identical texts across splits
texts_train = set(df_train["text"].str.lower())
print("Validation texts also in train:", df_val["text"].str.lower().isin(texts_train).sum())
print("Test texts also in train:", df_test["text"].str.lower().isin(texts_train).sum())
```
